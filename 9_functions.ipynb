```python
def inte(a):
    b=[]
    for i in a:
        if type(i)==int:
            b.append(i)
    return b        
```

```python
c=[1,7,8,6,.5,"amar","akbar",7,5,4]


inte(c)
```

```python
d=[1,7,9,0,"anthoni","Ganesh","ravi",True]


inte(d)
```

```python
def palindrome(strg):
    if strg==strg[::-1]:
        print("palindrome")

    else:
        print("not a palindrome")
```

```python
palindrome("madam")
```

```python
palindrome("amar")
```

```python
palindrome("343")
```
