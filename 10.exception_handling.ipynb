- Exceptions

exceptions are raised by during the python runtime

example

   - logic error (divide by 0)


   - Database Error 

```python
def divide(x,y):
    r=x/y
    print(r)
```

```python
divide(12,0)
```

this type of error we can handle with exception handling

how to handle this ?

   - try except block 

```python
def divide(x,y):
    if y==0:
        raise Exception ("we can't devide with zero")
    else:
        return x/y
```

```python
try:
    r=divide(10,0)
    print(r)
except Exception as e:
    print(e)    
```

```python
def divide(x,y):
    try:
        r=x/y
        print(r)
    except Exception as e:
        print(e)
```

```python
divide(10,0)
```
