```python
a=10

if a==12:
    print("a is equal to 12")
elif a==11:
    print("a is equal to 11")
elif a==10:
    print("a is equal to 10")

else:
    print("a is not equal to 12, 11 or 10")    
```

```python
a=10

if a==12:
    print("a is equal to 12")
elif a==11:
    print("a is equal to 11")
elif a!=10:
    print("a is equal to 10")

else:
    print("a is not equal to 12, 11 or 10") 
```
