# Class Methods

Class methods operate on the class itself, rather than on instances of the class. 

They are defined using the @classmethod decorator and take a cls parameter, which refers to the class, not the instance. They are used to access or modify class-level data, such as class attributes

```python
class netflix:
    new_show="kalki"


    def __init__(self,mail,password):
        self.mail=mail
        self.password=password
        self.service="Free"
        self.amount=0


    def get_password(self):
        return self.password


    @classmethod
    def get_newshow(cls):
        return cls.new_show


    @classmethod
    def change_newshow(cls,show_name):
        cls.new_show=show_name
```

```python
netflix.get_newshow()    # accessing of class method using classname 
```

```python
user=netflix("lakshman@gmail.com",1234)    # accesing class method using object
```

```python
user.get_newshow()
```

```python
netflix.change_newshow("salar")
                                    # modifying of classvariable using  class nane 
print(netflix.get_newshow())
```

```python
user=netflix("lakshman@gmail.com",1234) 


user.change_newshow("kalki-2")                      # modifying of classvariable using object


user.get_newshow()
```

# Abstraction
it is the process of hiding the internal implementation details and exposing only the essential functionality to the user

generally for impplementation of this using ABC and @abstractmethod

```python
from abc import ABC,abstractmethod


class atm(ABC):
    @abstractmethod
    def withdraw_money(self, amount):
        pass


class sbi_atm(atm):

    def withdraw_money(self, amount):
        print(f"verifying account...")
        print(f"processsing of withdrawl") 
        print(f" ${amount} withdraw successfully")   



obj=sbi_atm()

obj.withdraw_money(50000)
```

# single inheritance

one base class and one derived class , here transfer the properties parent class to child class



```python
class person():


    def __init__(self,name ,idnumber):
        self.name =name
        self.idnumber=idnumber

    def display(self):
        print(self.name)
        print(self.idnumber)


    def details(self):
        print("my name is ",self.name)
        print("my id number is ",self.idnumber)



class employee(person):

    def __init__(self, name, idnumber,salary, post):
        self.salary=salary
        self.post=post


        person.__init__(self, name,idnumber) 

          #     or

        #super().__init__(self,name, idnumber)
          

    def details1(self):
        print("my name is ",self.name)
        print("my id number is ",self.idnumber)
        print("post",self.post)




# creating obj

obj= employee("suhas",6987860,60000,"dentist")


        

obj.details()
print() 


obj.details1()
```

multiple inheritance:

```python
class father():
    def __init__(self,id):
        self.id=id


    def details(self):
        print("this is my id number",self.id)




class mother():
    def __init__(self,name):
        self.name=name

    def show(self):
        print("this is my name ",self.name)



class child(father,mother):
    def __init__(self, id,name,salary):
        father.__init__(self,id)
        mother.__init__(self,name)
        self.salary=salary


    def employee_details(self):
        print("this is my id number",self.id)
        print("this is my name ",self.name)
        print("this is my salary",self.salary)



# creating obj

obj=child(222,"suhas",75000)


obj.employee_details()

print()

obj.details()

print()

obj.show()  
```

# Multilevel inheritane:


```python
class Grandparent:

    def __init__(self,grandparent_name):
        self.grandparent_name =grandparent_name


    def display_grandparent(self):
        print("grand parent:",self.grandparent_name)



class parent(Grandparent):

    def __init__(self, grandparent_name,parent_name):
        super().__init__(grandparent_name)
        self.parent_name=parent_name


    def display_parent(self):
        print("parent:",self.parent_name)




class child(parent):

    def __init__(self, grandparent_name, parent_name,child_name):
        super().__init__(grandparent_name, parent_name) 
        self.child_name=child_name



    def display_child(self):
        print("child:",self.child_name)


# creating obj

obj=child("ram","rahim","akbar") 



# calling all methods with obj

obj.display_grandparent()

obj.display_parent()

obj.display_child()
```

# Hierarchical inheritance:

in hierarchical inheritance ,it has one parents has two or more childs that is called Hierarchical inheritance

```python
class parent():

    def __init__(self,id):
        self.id=id

    def details(self):
        print("this is my id number :" ,self.id)



class child1(parent):

    def __init__(self, id,name):
        super().__init__(id)
        self.name=name

    def employee_details(self):
        print("this is id number:",self.id)
        print("this is name:",self.name)


class child2(parent):

    def __init__(self, id,salary):
        super().__init__(id)
        self.salary=salary


    def employee_details1(self):
        print("this is id number :",self.id)
        print(" this is salary:",self.salary)


# creating obj for child1

obj=child1(222,"ravi")

obj.employee_details()

print()
# creating obj for child 2

obj1=child2(2223,70000)

obj1.employee_details1()   
```

# Hybrid inheritance:

- Hybrid inheritance in Python is a combination of two or more types of inheritance,

- such as combining hierarchical inheritance with multiple inheritance

- This structure often forms what is known as the "diamond problem"

```python
class school():
    def fun1(self):
        print("this is school function")



class student1(school):
    def fun2(self):
        print("this is student1 function")



class student2(school):
    def fun3(self):
        print("this is student2 function")


class student3(student1,school):
    def fun4(self):
        print("this is student3 fundction") 



# creating obj
obj=student3()


obj.fun1()
print()
obj.fun2()
print()
obj.fun4()
```
