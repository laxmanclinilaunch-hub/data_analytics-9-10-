# Class

A class is a blueprint for creating objects. It defines a set of attributes and methods that the created objects will have

Attributes: Variables that hold data related to the class.

Methods: Functions defined within a class that can operate on its attributes.

class is having attributes(variables) and behaviuor(methods) ,simply data + functions is called class

# Object

An object is an instance of a class. It is a specific realization of the class with actual values for the attributes.

```python
class person:

    a="school"       # class variables

    def __init__(self,name , age) :   # constructor
        self.name = name        # instance variables
        self.age =age 
        print(f"to chekc values are initilised or not {self.name} and {self.age}")
        


    def get_name(self):   # instance method
        return self.name 
    
    def get_age(self):
        return self.age
```

```python
obj= person("atul",25)
```

```python
obj.a    # accessing class name
```

```python
obj.age    # accessing instance variable
```

```python
obj.name      # accessing instance variable
```

```python
obj.get_age()      # accessing of instance methods
```

```python
obj.get_name()              # accessing of instance method
```
