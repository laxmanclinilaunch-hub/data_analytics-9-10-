# Class

1.Parameterized Constructor:

A parameterized constructor takes one or more arguments in addition to self . It allows for the initialization of an object with specific values provided at the time of object creation

2.Non-Parameterized Constructor

Initializes attributes with default values. It is useful when you want to provide default settings or when initialization does not

```python
class netflix:

    new_show = 22

    def __init__(self,mail, password,service,amount ): # 1.Parameterized Constructor
        self.mail = mail
        self.password = password
        self.service = service
        self.amount =  amount 

        


    def get_password(self):
        return self.service, self.amount
```

in the above example we are defined that parameters within __init__ and the same values are defined below as instance variables is called paramererized construtor

# Instance Methods
Instance methods are the most common methods in a class. They operate on instance data and can access and modify instance attributes.
 They are defined with at least one parameter, typically named self , which refers to the instance on

```python
class car() :
    def __init__(self , milage, year , make , model):
        self.milage = milage
        self.year = year
        self.make = make
        self.model = model

    def car_m(self):
      print("milage:",self.milage)

      print("year:",self.year)

    def car_details(self):
      print("milage:",self.milage)
      print("year;",self.year)
      print("model:",self.model)
```

```python
obj= car(20,2000,"honda","city")
```

```python
obj.car_m()
```

```python
obj.car_details()
```

# Class Methods

Class methods operate on the class itself, rather than on instances of the class. 

They are defined using the @classmethod decorator and take a cls parameter, which refers to the class, not the instance. They are used to access or modify class-level data, such as class attributes

```python
class netflix:
    new_show="kalki"


    def __init__(self,mail,password):
        self.mail=mail
        self.password=password
        self.service="Free"
        self.amount=0


    def get_password(self):
        return self.password


    @classmethod
    def get_newshow(cls):
        return cls.new_show


    @classmethod
    def change_newshow(cls,show_name):
        cls.new_show=show_name
```

```python
netflix.get_newshow()    # accessing of class method using classname 
```

```python
user=netflix("lakshman@gmail.com",1234)    # accesing class method using object
```

```python
user.get_newshow()
```

```python
netflix.change_newshow("salar")
                                    # modifying of classvariable using  class nane 
print(netflix.get_newshow())
```

```python
user=netflix("lakshman@gmail.com",1234) 


user.change_newshow("kalki-2")                      # modifying of classvariable using object


user.get_newshow()
```

# - encapsulation:
it is a pillar of oops

it using varibles and methods we have to controlling access to the data

a core concept in programming where data and the functions that use that data are bundled together into a single unit

# public memeber:

it can access anywhere (inside the class or outside of the class)

   - self.name=name
   
# proteced member:

the protected memeber start with one underscore

it can access inside the class and its subclass ( child class)

 - self._name = name 
# private memeber

private memeber start with duble underscore

the private memeber can't access directly

to access private member we have to (getter) method and for modification use (setter) method

        - self.__name = name 

- public member


```python
class mentor:
    def __init__(self):
        self.name="laxman"
        


obj=mentor()

print(obj.name)
```

- protected member

```python
class mentor:
    def __init__(self):
        self._name='laxman'
        

obj=mentor()

print(obj._name)  # protected member
```

```python
class father:
    def __init__(self):
        self._salary=70000




class son(father):
    def display(self):
        print(self._salary)



obj=son()


obj.display()
```

- example

```python
class patient:
    def __init__(self):
        self.name="arunachalam"   # public member 
        self._blood_group="A1 NEGATIVE"   # protected member 
        self.__medical_report="HIV"     # private

    def get_medical_report(self):
        return self.__medical_report


obj=patient()


print(obj.name)   # accessing of public member
print(obj._blood_group)   # accessing of protected member
print(obj.get_medical_report())  # accessing of private member
```

# Abstraction
it is the process of hiding the internal implementation details and exposing only the essential functionality to the user

generally for impplementation of this using ABC and @abstractmethod

```python
from abc import ABC,abstractmethod


class atm(ABC):
    @abstractmethod
    def withdraw_money(self, amount):
        pass


class sbi_atm(atm):

    def withdraw_money(self, amount):
        print(f"verifying account...")
        print(f"processsing of withdrawl") 
        print(f" ${amount} withdraw successfully")   



obj=sbi_atm()

obj.withdraw_money(50000)
```
