- it is python file (.py)

- this file contains functions , variables , classes that you want to include another program file

- for use module , we can use import in other file (scripts)

example math_utilis.py

- # Package

- Package is directory , that contains multiple modules or combination of multiple modules
- the package has a special **__ init __.py**  file

- collection of packages and modules is called library

# import specific function or class

- from module_name import function function_name

```python
from math import sqrt

sqrt(25)
```

# - import multiple functions or classes

- from module_name import function1 , function2 ...

```python
from math import sqrt, pi

print(pi)
```

# - import all functions or classes from a module

- from module_name import *

```python
from math import *

sqrt(16)
```

- IMPORT FROM A SUBMODULE OR PACKAGE

     - from package_name import module_name
