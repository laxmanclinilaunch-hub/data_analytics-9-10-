- Break ,  continue , pass


- Break:
   - when specific condition met, break allows to terimnate(stop) the loop

```python
a="python"

for i in a:
    if i=='h':
        break
    print(i)
```

```python
b=[1,2,3,4,5]

for j in b:
    if j==3:
        break
    print(j)
```

- continue:

   - when specific condition met, continue allows to skipping some part of loop and control transfered to the next loop 

```python
a="python"

for i in a:
    if i=='h':
        continue
    print(i)
```

```python
b=[1,2,3,4,5,6,3,7,8,9]

for j in b:
    if j==3:
        continue
    print(j)
```
