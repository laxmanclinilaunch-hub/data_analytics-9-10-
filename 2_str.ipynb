# Indexing
  - index means finds the location of element
  - two types
    -  positive/ forward index:
         - it starts from left to right 
         - it element numbers start from 0 , next position 1  and so on 2,3 ,4.....
    - negative / backward index:
         - it starts from right to left 
         - it element numbers start from -1 , next position -2, and so on -3,-4,-5 ......
