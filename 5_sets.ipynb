# - sets

- sets are an unordered collection of unique elements

- we can construct the sets by using set() function

- sets are mutable just like lists

- for creation of Non-empty set using { } curly braces and each element separated commas

```python
empty_set=set()

empty_set
```

```python
type(empty_set)
```

```python
non_empty_set={"amar","afiz khan","ganesh",3,5.5}


print(non_empty_set)
```
